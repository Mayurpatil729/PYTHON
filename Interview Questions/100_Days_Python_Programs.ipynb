### 1. Take 3 ages from user and find Oldest one

```python
Age1=int(input("Enter Age1:"))
Age2=int(input("Enter Age2:"))
Age3=int(input("Enter Age3:"))
if Age1>Age2 and Age1>3:
    print(f"Age1")
elif Age2>Age1 and Age2>Age3:
    print(f"Age2")
else:
    print(f"Age3")
```
